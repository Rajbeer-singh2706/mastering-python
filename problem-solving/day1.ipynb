## Day 1 — FizzBuzz & Number Patterns

### Problem set (3 levels)

**Level 1 — Classic FizzBuzz**
Print numbers 1–100. But for multiples of 3 print `"Fizz"`, multiples of 5 print `"Buzz"`, and multiples of both print `"FizzBuzz"`.

**Level 2 — Generalised FizzBuzz**
Write a version where the rules (`divisor → word`) are passed in as a dictionary, so the caller can define any mapping without changing your core logic.

````python
### Starter code

# ── Level 1: Classic FizzBuzz ──────────────────────────────────────
def fizzbuzz(n: int) -> list[str]:
    result = []
    for i in range(1, n + 1):
        word = ""
        if i % 15 == 0:
            word = "FizzBuzz"
        elif i % 3 == 0:
            word = "Fizz"
        elif i % 5 == 0:
            word = "Buzz"
        result.append(word or str(i))
    return result

print(fizzbuzz(20))
````

````python
# ── Level 2: Generalised FizzBuzz ─────────────────────────────────
def fizzbuzz_general(n: int, rules: dict[int, str]) -> list[str]:
    result = []
    for i in range(1, n + 1):
        # TODO: iterate rules in sorted order, build word
        word = ""
        for rule_id, rule_word in rules.items():
            if i % rule_id == 0:
                word += rule_word
        result.append(word or str(i))
    return result

rules = {3: "Fizz", 5: "Buzz", 7: "Bazz"}
print(fizzbuzz_general(21, rules))
````

````python
def print_triange(n: int) -> None:
    for i in range(1, n + 1):
        spce = " " * (n - i)
        print(spce, end="")
        for j in range(1, i + 1):
            print(f"{j}", end="")
        print()  # Newline after each row
print_triange(5)
````

# Hints (reveal one at a time — don't peek ahead!)






**Level 1**
- `i % 3 == 0` checks divisibility. Check *both* conditions first before checking each alone, or build a string by concatenating `"Fizz"` and `"Buzz"` conditionally — then you never need an explicit `"both"` branch.

**Level 2**
- Sort the dictionary keys before iterating: `for divisor in sorted(rules)`. This ensures `3` and `5` always combine into `"FizzBuzz"` correctly regardless of dict insertion order.

**Level 3 — Triangle**
- `"".join(str(d) for d in range(1, row+1))` builds `"123...row"`. Then `str.rjust(width)` right-pads with spaces — no manual counting needed.

**Level 3b — Diamond**
- Use `abs(row)` to mirror the top half onto the bottom. For the middle row (`row == 0`) print one `*`. For all other rows, the two stars sit at positions `abs(row)` from each edge — a single space of width `2*(n-1-abs(row))-1` goes between them.
