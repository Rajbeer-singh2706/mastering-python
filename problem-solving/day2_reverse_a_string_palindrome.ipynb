# Day 2 — String Reversal & Palindrome Check

```python
#Level 1 — Reverse a string
#Reverse a given string without using slicing ([::-1]). Use an explicit loop or built-in iteration.

def reverse(s: str) -> str:
    l = len(s) // 2
    length = len(s) - 1
    list_s = list(s)
    for i in range(l):
        list_s[i], list_s[length - i] = list_s[length - i], list_s[i]
    return ''.join(list_s)

result = reverse("bca")
result 
```

```python
# Level 2 — Palindrome check
# Check if a string is a palindrome. Ignore spaces, punctuation, and case.
```

```python
def is_palindrome(s: str) -> bool:
    s = ''.join(c for c in s if c.isalnum()).lower()
    right = len(s) - 1
    for left in range(len(s) //2 ):
        if s[left] != s[right]:
            return False
        right -= 1
    return True

result = is_palindrome("A man a plan a canal Panama")
print(result)
result = is_palindrome("hello")
print(result)
```
