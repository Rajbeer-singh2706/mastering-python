# Interview Questions



**Q1. What makes Python an interpreted language?**

> Code is executed line-by-line at runtime by the interpreter (CPython compiles to bytecode first, then interprets). No separate compile step needed.

**Q2. What is PEP 8?**

> Python's official style guide — 4 spaces indentation, 79-char line limit, naming conventions (snake_case for variables/functions, PascalCase for classes).


### 2. Receipt Printer
Print a formatted store receipt with 3 items, prices, subtotal, tax (18%), and grand total.
```python
# Your solution
```

````python
print("#" * 20)
item1_name = "Water"
item1_price = 20
item1_subtotal = item1_price + (item1_price*0.18)

item2_name = "Chapati"
item2_price = 150
item2_subtotal = item2_price + (item2_price*0.18)

item3_name = "Sabji"
item3_price = 200
item3_subtotal = item3_price + (item3_price*0.18)

print(f"Item1 : {item1_name}, prices: {item1_price},  tax(18%) : subtotal : {item1_subtotal}")
print(f"Item2 : {item2_name}, prices: {item2_price},  tax(18%) : subtotal : {item2_subtotal}")
print(f"Item3 : {item3_name}, prices: {item3_price},  tax(18%) : subtotal : {item3_subtotal}")

grand_total = item1_subtotal + item2_subtotal + item3_subtotal
print(f"Grand Total: {grand_total}")
print("#" * 20)
````

## 🚀 Use Case — System Info Dashboard

````python
import sys, platform
from datetime import datetime
print("="*50)
print("       SYSTEM HEALTH DASHBOARD")
print("="*50)
print(f"  Timestamp : {datetime.now():%Y-%m-%d %H:%M:%S}")
print(f"  Python    : {sys.version.split()[0]}")
print(f"  Platform  : {platform.system()} {platform.release()}")
print(f"  Machine   : {platform.machine()}")
print("="*50)
print("  Status    : ✅ All Systems Operational")
print("="*50)
````
