## 1. All Built-in Data Types

````python
age      = 28          # int
height   = 5.9         # float
z        = 2 + 3j      # complex
name     = "Python"    # str
active   = True        # bool
data     = None        # NoneType
l = [("age",age),("height",height),("z",z),("name",name),("active",active),("data",data)]
for var,val in l:
    print(f"  {var:8s} = {str(val):15s}  type={type(val).__name__}")
````

## 2. Dynamic Typing

````python
x = 10; 
print(f"x={x}, type={type(x).__name__}")
x = "now a string"; 
print(f"x={x!r}, type={type(x).__name__}")
x = [1,2,3]; 
print(f"x={x}, type={type(x).__name__}")
````

## 3. Type Conversion

````python
# Implicit
result = 5 + 2.0; 
print(result, type(result))  # float

# Explicit
print(int(3.99))          # 3  (truncate)
print(float("3.14"))      # 3.14
print(str(100))           # '100'
print(bool(0), bool(""))  # False False
print(bool(1), bool("hi"))# True True

# Pitfall
s = "42"
# print(s + 1)  ← TypeError!
print(int(s) + 1)         # 43 ✓
````

## 4. isinstance() vs type()

````python
x = True
print(type(x).__name__)      # bool
print(isinstance(x, bool))   # True
print(isinstance(x, int))    # True — bool IS-A int
# Always prefer isinstance() for type checks in production
````

## 🎤 Interview Questions

**Q1. Difference between 'is' and '=='?**

> == compares values. 'is' compares object identity (same memory address). [1] == [1] → True but [1] is [1] → False.

**Q2. What are truthy/falsy values?**

> Falsy: None, False, 0, 0.0, '', [], {}, set(). Everything else is truthy. Python uses this in boolean contexts.

## 🚀 Use Case — Record Type Validator

````python
def validate_record(record, schema):
    errors = []
    for field, expected in schema.items():
        if field not in record:
            errors.append(f"Missing: {field}")
        elif not isinstance(record[field], expected):
            errors.append(f"{field}: expected {expected.__name__}, got {type(record[field]).__name__}")
    if errors:
        print("❌ INVALID:"); [print(f"   • {e}") for e in errors]
    else:
        print("✅ VALID")

schema = {"user_id":int,"name":str,"score":float,"active":bool}
validate_record({"user_id":1,"name":"Alice","score":95.5,"active":True}, schema)
validate_record({"user_id":"2","name":"Bob","score":"88","active":True}, schema)
````
