```python
print("#" * 20)
item1_name = "Water"
item1_price = 20
item1_subtotal = item1_price + (item1_price*0.18)

item2_name = "Chapati"
item2_price = 150
item2_subtotal = item2_price + (item2_price*0.18)

item3_name = "Sabji"
item3_price = 200
item3_subtotal = item3_price + (item3_price*0.18)

print(f"Item1 : {item1_name}, prices: {item1_price},  tax(18%) : subtotal : {item1_subtotal}")
print(f"Item2 : {item2_name}, prices: {item2_price},  tax(18%) : subtotal : {item2_subtotal}")
print(f"Item3 : {item3_name}, prices: {item3_price},  tax(18%) : subtotal : {item3_subtotal}")

grand_total = item1_subtotal + item2_subtotal + item3_subtotal
print(f"Grand Total: {grand_total}")
print("#" * 20)
```
