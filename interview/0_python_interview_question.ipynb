```python

input = ["123","321","213","456"]
#-- Output - [["123","321","213"],["456"]]

anagram_data = {}


for item in input:
    key = "".join(sorted(item))
    if key not in anagram_data:
        anagram_data[key] = [item]
    else:
        anagram_data[key].append(item)

print(list(anagram_data.values()))
```
